```python
from tensorflow.keras.preprocessing.text import one_hot
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding
import numpy as np
from tensorflow.keras.utils import pad_sequences
```

```python
sent = [
    'The rusty anchor scraped against the coral reef as the tide began to turn',
    'She poured a generous amount of maple syrup over her stack of blueberry pancakes',
    'Neon lights flickered in the rain puddles outside the deserted diner',
    'A sudden gust of wind scattered the autumn leaves across the freshly cut lawn',
    'He carefully adjusted the telescope lens to focus on the distant crater of the moon',
    'The old grandfather clock in the hallway chimed three times before stopping completely',
    'Vibrant murals covered every inch of the concrete wall under the highway overpass'
]
```

```python
one_hot_representation = [one_hot(words,10000) for words in sent]
```

```python
max_len = 20
sent_padded = pad_sequences(one_hot_representation,maxlen=max_len,padding='pre')
```

```python
dim=10
```

```python
model = Sequential()
model.add(Embedding(10000,10,input_length=max_len))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
result = model.predict(sent_padded[0])
```
